# **Business Problem Understanding**

## **Context**

Pembatalan reservasi yang mendadak dapat menurunkan pendapatan dan menimbulkan kerugian pada hotel, karena kamar yang sudah dipesan menjadi kosong dan sulit diisi kembali dalam waktu singkat. Untuk mengantisipasi hal ini, hotel ingin memanfaatkan data historis, seperti informasi pelanggan, tipe kamar, segmen pasar, dan variabel terkait lainnya, untuk memprediksi kemungkinan pembatalan reservasi. Dengan mengetahui reservasi yang berisiko tinggi dibatalkan, hotel dapat mengambil langkah preventif guna meminimalkan dampak negatif.

Target prediksi: `is_canceled`
* 0: Tidak dibatalkan
* 1: Dibatalkan

## **Problem Statement**

Proses pembatalan reservasi dapat berdampak langsung pada pendapatan hotel. Tanpa adanya informasi atau prediksi mengenai kemungkinan pembatalan, hotel akan kesulitan dalam mengelola tingkat hunian kamar secara efisien dan merancang strategi untuk mengantisipasi pembatalan mendadak, selain itu hotel juga gagal memperoleh 100% pendapatan dari reservasi tersebut. Hotel ingin memprediksi kemungkinan pembatalan reservasi agar dapat mengurangi potensi kerugian dan meningkatkan efisiensi operasional. Dengan memanfaatkan data yang tersedia, seperti asal negara pelanggan, segmen pasar, riwayat pembatalan sebelumnya, perubahan dalam reservasi, serta permintaan khusus, hotel berharap dapat membuat keputusan yang lebih baik sebelum pembatalan terjadi.

Contoh Kasus: <br>
Seorang pelanggan memesan kamar seharga IDR 400k per malam untuk satu malam. Jika pelanggan membatalkan reservasi tersebut, hotel akan kehilangan pendapatan IDR 400k dan tetap menanggung biaya operasional. Selain itu, hotel perlu mengalokasikan waktu dan sumber daya untuk mengisi kembali kamar kosong tersebut, yang tidak selalu mudah jika waktu pembatalan terlalu mendadak. Oleh karena itu, kemampuan untuk memprediksi pembatalan akan sangat membantu hotel dalam mengambil tindakan antisipatif, seperti memberikan promo khusus atau menyesuaikan strategi pemasaran.

## **Goals**

Hotel ingin memiliki kemampuan untuk memprediksi kemungkinan pembatalan setiap reservasi yang masuk. Informasi ini akan membantu tim pemasaran untuk lebih fokus pada pelanggan yang berisiko tinggi membatalkan reservasi dan mengambil langkah preventif yang tepat, seperti memberikan penawaran khusus, diskon, atau pengingat yang dipersonalisasi, guna mengurangi kemungkinan pembatalan.

Selain itu, hotel juga ingin memahami faktor-faktor utama yang memengaruhi keputusan pelanggan dalam membatalkan reservasi. Wawasan ini akan menjadi dasar bagi tim pemasaran dalam merancang strategi promosi dan kebijakan harga yang lebih efektif, serta menciptakan pengalaman pelanggan yang lebih baik guna meningkatkan loyalitas dan menurunkan tingkat pembatalan di masa mendatang.

## **Analytic Approach**

Hotel akan menganalisis data historis untuk menemukan pola yang membedakan antara reservasi yang dibatalkan dan yang tidak dibatalkan. Berdasarkan pola tersebut, akan dibangun model klasifikasi yang dapat memprediksi kemungkinan pembatalan suatu reservasi dengan mempertimbangkan berbagai fitur seperti tipe kamar, segmen pasar, status deposit, dan informasi lainnya. Model ini diharapkan dapat membantu tim pemasaran dalam mengambil langkah proaktif untuk mengurangi tingkat pembatalan.

Keuntungan untuk Tim Pemasaran Hotel (Stake Holder):
* **Menargetkan Pelanggan dengan Risiko Tinggi**: Model prediksi memungkinkan tim pemasaran mengidentifikasi pelanggan yang berpotensi tinggi untuk membatalkan reservasi. Dengan informasi ini, tim dapat menyusun strategi komunikasi atau promosi yang tepat sasaran.
* **Pemberian Insentif dan Promo Tepat Waktu**: Dengan mengetahui kemungkinan pembatalan lebih awal, tim pemasaran dapat menawarkan diskon, upgrade kamar, atau benefit lain yang dapat meningkatkan komitmen pelanggan untuk tetap melanjutkan reservasi.
* **Optimalisasi Anggaran Pemasaran**: Sumber daya pemasaran dapat difokuskan pada segmen pelanggan yang benar-benar membutuhkan perhatian, sehingga strategi menjadi lebih efisien dan berdampak.

## **Metric Evaluation**

**False Positive** (Type 1 Error): 
* Model secara keliru memprediksi bahwa sebuah reservasi akan dibatalkan, padahal sebenarnya tidak. Akibatnya, hotel mungkin memberikan diskon atau insentif yang tidak perlu. Meskipun menambah biaya, dampaknya masih lebih kecil dibandingkan kehilangan pendapatan dari kamar kosong.

**False Negative** (Type 2 Error):
* Model gagal mendeteksi reservasi yang sebenarnya akan dibatalkan. Hal ini lebih merugikan karena hotel kehilangan pendapatan penuh dari kamar yang tidak jadi dihuni, dan tidak sempat mengambil tindakan pencegahan.

Metrik utama yang digunakan adalah **F2 Score**, karena metrik ini tetap mempertimbangkan Precision namun memberikan **bobot lebih besar pada Recall**. Dalam konteks prediksi pembatalan reservasi, kesalahan yang paling merugikan adalah ketika model gagal mendeteksi pembatalan (False Negative), karena hotel kehilangan peluang untuk menjual ulang kamar. Dengan menggunakan F2 Score, model didorong untuk lebih sensitif dalam mengenali potensi pembatalan tanpa sepenuhnya mengabaikan ketepatan prediksi (Precision), sehingga hotel dapat mengantisipasi lebih banyak pembatalan dan mengurangi kerugian pendapatan.

```python
def summary_data(df:pd.DataFrame, expand_row:bool=False) -> pd.DataFrame:
    """
    Generate summary of a DataFrame with the following details per column:
    - Data type
    - Null count and percentage
    - Number of unique values
    - Sample of unique values
    """
    pd.set_option('display.max_colwidth', None if expand_row else 50)

    total_rows = len(df)

    summary = {
        "Column Name": [],
        "Type": [],
        "Null": [],
        "Null (%)": [],
        "Number of Unique": [],
        "Unique Sample": []
    }

    for col in df.columns:
        col_data = df[col]
        null_count = col_data.isna().sum()

        summary["Column Name"].append(col)
        summary["Type"].append(col_data.dtype)
        summary["Null"].append(null_count)
        summary["Null (%)"].append(round(null_count / total_rows * 100, 2))
        summary["Number of Unique"].append(col_data.nunique())
        summary["Unique Sample"].append(col_data.unique())

    return pd.DataFrame(summary)
```

```python
def distribution_check(x,capt:str, bins="auto"):

    fig, axes = plt.subplots(1,2, figsize=(12, 3))

    sns.histplot(x=x, bins="auto", kde=True, ax=axes[0])
    axes[0].set_title(f"Histogram {capt}")

    sns.boxplot( x=x, ax=axes[1])
    axes[1].set_title(f"Box plot {capt}")

    plt.show()

    stat, pvalue = normaltest(x)
    print(f'Distribusi Normal P-Val={pvalue:.2f}' if pvalue > 0.05 else f'Distribusi Tidak Normal P-Val={pvalue:.2f}')
```

```python
def find_rare_values(df, col, threshold=0.05):
    freq = df[col].value_counts(normalize=True)
    count = df[col].value_counts()
    
    rare = freq[freq < threshold]
    
    result = pd.DataFrame({
        "Kolom": col,
        "Value": rare.index,
        "Count": count[rare.index].values,
        "Frequency": rare.values
    })

    return result


def find_rare_values_in_df(df, threshold=0.05):
    rare_result = pd.DataFrame()

    for col in df.select_dtypes(include="object").columns:
        rare_values = find_rare_values(df, col, threshold=threshold)
        if not rare_values.empty:
            rare_result = pd.concat([rare_result, rare_values], ignore_index=True)

    return rare_result
```

## **Imbanlace Data Experiment**

```python
def cross_val_models(X_train, y_train, models, encoder=encoder, scaler=scaler, cv_folds=5, random_state=42):
    results = []
    cv = StratifiedKFold(n_splits=cv_folds, shuffle=True, random_state=random_state)
    f2 = make_scorer(fbeta_score, beta=2, zero_division=0)

    for name, model in models:
        # Apply scaling only for models sensitive to feature magnitude
        if name in ['Logistic Regression', 'KNN Classifier', 'SVM Classifier']:
            pipeline = Pipeline([
                ('encoder', encoder),
                ('scaler', scaler),
                ('model', model)
            ])
        else: # tree based gk pake scaler
            pipeline = Pipeline([
                ('encoder', encoder),
                ('model', model)
            ])

        scores = cross_validate(
            pipeline,
            X=X_train,
            y=y_train,
            scoring=f2,
            cv=cv,
            return_train_score=False
        )

        results.append({
            'Model': name,
            'Mean F2': np.round(scores['test_score'].mean(), 3),
            'Std Dev': np.round(scores['test_score'].std(), 3),
            'F2 Scores': np.round(scores['test_score'], 3),
            'Fit Time': np.round(scores['fit_time'].mean(), 3),
            'Score Time': np.round(scores['score_time'].mean(), 3),
        })

    return pd.DataFrame(results).sort_values(by='Mean F2', ascending=False).reset_index(drop=True)
```

## **Balance Data Experiment**

**Pada eksperimen kedua ini, fokus diarahkan pada penanganan ketidakseimbangan data (imbalance) menggunakan berbagai teknik resampling.** Beberapa metode yang akan diterapkan meliputi:

* **Over Sampling**

  * **SMOTE (Synthetic Minority Over-sampling Technique)**:  
    Membuat data sintetis dari kelas minoritas dengan menginterpolasi antar data yang ada.

  * **ADASYN (Adaptive Synthetic Sampling)**:  
    Menghasilkan data sintetis dengan fokus pada contoh yang sulit diklasifikasikan.

* **Under Sampling**

  * **Random UnderSampler**:  
    Mengurangi jumlah data dari kelas mayoritas secara acak untuk mencapai keseimbangan.

  * **NearMiss (Versi 3)**:  
    Memilih contoh dari kelas mayoritas yang paling dekat dengan data kelas minoritas.

* **Hybrid Sampling**

  * **SMOTEENN (SMOTE + Edited Nearest Neighbours)**:  
    Kombinasi antara SMOTE dan metode pembersihan data ENN, yang menghapus sampel mayoritas yang ambigu setelah oversampling.


Dengan pendekatan ini, eksperimen bertujuan untuk meningkatkan performa model pada data yang tidak seimbang.


```python
resamplers = [
    # Undersampling
    ('RandomUnderSampler', RandomUnderSampler(random_state=42)),
    ('NearMiss v3', NearMiss(version=3)),

    # Oversampling
    ('SMOTE', SMOTE(random_state=42)),
    ('ADASYN', ADASYN(random_state=42)),

    # Hybrid (SMOTE + cleaning)
    ('SMOTEENN', SMOTEENN(random_state=42))
]
```

```python
def cross_val_bal(X_train, y_train, model_name, model, resamples=resamplers, encoder=encoder, scaler=scaler, cv_folds=5, random_state=42):
    results = []
    cv = StratifiedKFold(n_splits=cv_folds, shuffle=True, random_state=random_state)
    f2 = make_scorer(fbeta_score, beta=2, zero_division=0)

    for name, resample in resamples:
        # Apply scaling only for models sensitive to feature magnitude
        if name in ['Logistic Regression', 'KNN Classifier', 'SVM Classifier' ]:
            pipeline = Pipeline([
                ('encoder', encoder),
                ('scaler', scaler),
                ('resample', resample),
                ('model', model)
            ])
        else:  # tree-based models, no scaling
            pipeline = Pipeline([
                ('encoder', encoder),
                ('resample', resample),
                ('model', model)
            ])

        scores = cross_validate(
            pipeline,
            X=X_train,
            y=y_train,
            scoring=f2,
            cv=cv,
            return_train_score=False
        )

        results.append({
            'Model': model_name,
            'Resample': name,
            'Mean F2': np.round(scores['test_score'].mean(), 3),
            'Std Dev': np.round(scores['test_score'].std(), 3),
            'F2 Scores': np.round(scores['test_score'], 3),
            'Fit Time': np.round(scores['fit_time'].mean(), 3),
            'Score Time': np.round(scores['score_time'].mean(), 3),
        })

    return pd.DataFrame(results).sort_values(by='Mean F2', ascending=False).reset_index(drop=True)
```

```python
balance_result = pd.DataFrame()
for name, model in tqdm(models, desc="Cross-validating models", leave=True):
    result = cross_val_bal(x_train, y_train, model_name=name, model=model, resamples=resamplers)
    balance_result = pd.concat([balance_result, result])
```

**A. Biaya Terkait Pembatalan Reservasi Hotel**

Untuk mengevaluasi manfaat dari model machine learning dalam konteks prediksi pembatalan reservasi hotel, kita akan membandingkan biaya tanpa model (baseline) dan biaya dengan model. Fokus utama adalah meminimalkan kerugian akibat pembatalan melalui intervensi dini (misalnya, diskon untuk mencegah pembatalan).

**B. Asumsi**

* Harga sewa kamar per malam: **Rp 400.000**
  ([Sumber](https://www.budgetyourtrip.com/hotels/indonesia/jakarta-1642911))
* Diskon pencegahan pembatalan: **10%**
* Dengan ML, kita bisa menawarkan diskon hanya kepada pelanggan **yang diprediksi akan membatalkan** (positif)

**C. Simulasi Biaya**

**1. Tanpa Machine Learning** (semua pembatalan tidak dapat dicegah)

  Pembatalan terjadi pada semua pelanggan aktual yang batal (TP + FN = **450 + 15 = 465**):

  $$
  465 \text{ pembatalan} × Rp 400.000 = \boxed{Rp 186.000.000} \text{ kerugian total}
  $$

**2. Dengan Machine Learning**

  * **a. Intervensi diberikan ke semua yang diprediksi akan batal:**

    Model memprediksi **1.090 pelanggan** akan membatalkan (TP + FP = 450 + 640), dan semuanya diberi diskon 10% sebagai upaya pencegahan.

    * **TP (450)** – batal & berhasil dicegah →
      Pendapatan masuk (setelah diskon):
      **450 × Rp 400.000 × 90% = Rp 162.000.000**

    * **FP (640)** – tidak batal tapi tetap diberi diskon →
      Biaya diskon sia-sia:
      **640 × Rp 400.000 × 10% = Rp 25.600.000**

    * **FN (15)** – batal & tidak dicegah →
      Rugi penuh:
      **15 × Rp 400.000 = Rp 6.000.000**

  * **b. Total kerugian dengan ML model:**

    * Pendapatan diselamatkan: **Rp 162.000.000**
    * Kerugian: FP (diskon sia-sia) + FN (batal tidak dicegah) =

      $$
      Rp 25.600.000 + Rp 6.000.000 = \boxed{Rp 31.600.000}
      $$


  **D. Perbandingan Total Biaya**

  | Skenario      | Kerugian Total               |
  | ------------- | ---------------------------- |
  | Tanpa ML      | Rp 186.000.000               |
  | Dengan ML     | Rp 31.600.000                |
  | **Efisiensi** | **↓ Rp 154.400.000** (≈ 83%) |


**E. Kesimpulan**

Dengan memanfaatkan model machine learning dan memberikan intervensi berupa diskon 10% kepada pelanggan berisiko, potensi kerugian bisnis bisa ditekan dari **Rp 186 juta menjadi hanya Rp 31,6 juta**, menghemat sekitar **83%**. Ini menunjukkan bahwa model sangat bermanfaat secara finansial, meskipun tetap ada biaya dari diskon kepada pelanggan yang sebenarnya tidak akan membatalkan (false positives).

# **Shapley Value of Explanations (SHAP)**

```python
# Ambil model final dari pipeline
shap_model = best_model.named_steps["model"]

# Transformasi X (sampai sebelum ADASYN dan model)
X_transformed = best_model[:-2].transform(x_test)
# Ambil nama fitur dari encoder
encoder = best_model.named_steps["encoder"]
feature_names = encoder.get_feature_names_out()

# Konversi jadi DataFrame agar SHAP bisa baca nama fitur
if hasattr(X_transformed, "toarray"):  # jika hasilnya sparse matrix
    X_transformed = X_transformed.toarray()

X_df = pd.DataFrame(X_transformed, columns=feature_names)

# Ambil sample representatif 
X_sample = X_df.sample(n=100, random_state=42)
X_sample_array = X_sample.values

# Buat explainer untuk model
explainer = shap.Explainer(shap_model.predict, X_sample_array)
shap_values = explainer(X_sample_array)

# Visualisasi SHAP (pakai nama fitur dari DataFrame agar tetap informatif)
shap.plots.beeswarm(shap.Explanation(values=shap_values.values,
                                        data=X_sample_array,
                                        feature_names=feature_names))
```
